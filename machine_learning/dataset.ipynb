```python
# ============================================
# Data Cleaning - Step 4
# Standardize Column Names
# ============================================

def clean_column_names(df):
    df.columns = (
        df.columns
        .str.strip()                  # Remove leading/trailing spaces
        .str.lower()                  # Convert to lowercase
        .str.replace(" ", "_", regex=False)
        .str.replace("-", "_", regex=False)
        .str.replace("%", "percent", regex=False)
        .str.replace("/", "_", regex=False)
        .str.replace("(", "", regex=False)
        .str.replace(")", "", regex=False)
    )
    return df

# Apply to all datasets
body = clean_column_names(body)
health = clean_column_names(health)
injury = clean_column_names(injury)
activity = clean_column_names(activity)
sleep = clean_column_names(sleep)
heart = clean_column_names(heart)
weight = clean_column_names(weight)

print("✅ Column names standardized successfully!")
```

```python
# ============================================
# Feature Engineering - Step 1
# BMI Category
# ============================================

def bmi_category(bmi):

    if bmi < 18.5:
        return "Underweight"

    elif bmi < 25:
        return "Normal"

    elif bmi < 30:
        return "Overweight"

    else:
        return "Obese"


weight["bmi_category"] = weight["bmi"].apply(bmi_category)

print("✅ BMI Category Created Successfully!\n")

print(weight[["bmi", "bmi_category"]].head())
```

```python
# ============================================
# Feature Engineering - Step 2
# Blood Pressure Category
# ============================================

def bp_category(systolic, diastolic):

    if systolic < 120 and diastolic < 80:
        return "Normal"

    elif 120 <= systolic < 130 and diastolic < 80:
        return "Elevated"

    elif (130 <= systolic < 140) or (80 <= diastolic < 90):
        return "Hypertension Stage 1"

    elif systolic >= 140 or diastolic >= 90:
        return "Hypertension Stage 2"

    else:
        return "Unknown"


body["bp_category"] = body.apply(
    lambda row: bp_category(row["systolic"], row["diastolic"]),
    axis=1
)

print("✅ Blood Pressure Category Created Successfully!\n")

print(body[["systolic", "diastolic", "bp_category"]].head())
```

```python
print(body["bp_category"].value_counts())
```

```python
# ============================================
# Feature Engineering - Step 3
# Sleep Efficiency
# ============================================

sleep["sleep_efficiency"] = (
    sleep["total_minutes_asleep"] / sleep["total_time_in_bed"]
) * 100

sleep["sleep_efficiency"] = sleep["sleep_efficiency"].round(2)

print("✅ Sleep Efficiency Created Successfully!\n")

print(
    sleep[
        [
            "total_minutes_asleep",
            "total_time_in_bed",
            "sleep_efficiency"
        ]
    ].head()
)
```

```python
# ============================================
# Sleep Quality Category
# ============================================

def sleep_quality(efficiency):

    if efficiency >= 90:
        return "Excellent"

    elif efficiency >= 85:
        return "Good"

    elif efficiency >= 75:
        return "Average"

    else:
        return "Poor"


sleep["sleep_quality"] = sleep["sleep_efficiency"].apply(sleep_quality)

print("\n✅ Sleep Quality Category Created!\n")

print(
    sleep[
        [
            "sleep_efficiency",
            "sleep_quality"
        ]
    ].head()
)
```

```python
print(sleep["sleep_quality"].value_counts())
```

```python
# ============================================
# Feature Engineering - Step 4
# Activity Level
# ============================================

def activity_level(steps):

    if steps < 5000:
        return "Sedentary"

    elif steps < 7500:
        return "Low Active"

    elif steps < 10000:
        return "Moderately Active"

    else:
        return "Highly Active"


activity["activity_level"] = activity["total_steps"].apply(activity_level)

print("✅ Activity Level Created Successfully!\n")

print(activity[["total_steps", "activity_level"]].head())

print("\nActivity Level Distribution:\n")

print(activity["activity_level"].value_counts())
```

```python
# ============================================
# Feature Engineering - Step 5
# Heart Rate Summary
# ============================================

heart_summary = (
    heart.groupby("id")
         .agg(
             avg_heart_rate=("value", "mean"),
             min_heart_rate=("value", "min"),
             max_heart_rate=("value", "max")
         )
         .reset_index()
)

# Round average heart rate
heart_summary["avg_heart_rate"] = heart_summary["avg_heart_rate"].round(2)

print("✅ Heart Rate Summary Created Successfully!\n")

print(heart_summary.head())

print("\nShape of Heart Summary Dataset:")
print(heart_summary.shape)
```

```python
# ============================================
# Feature Engineering - Step 6
# Heart Rate Category
# ============================================

def heart_rate_category(avg_hr):

    if avg_hr < 60:
        return "Low"

    elif avg_hr <= 100:
        return "Normal"

    else:
        return "High"


heart_summary["heart_rate_category"] = heart_summary["avg_heart_rate"].apply(
    heart_rate_category
)

print("✅ Heart Rate Category Created Successfully!\n")

print(heart_summary)

print("\nCategory Distribution:\n")

print(heart_summary["heart_rate_category"].value_counts())
```

```python
# ============================================
# Feature Engineering - Step 7
# Prepare Fitbit Datasets for Merge
# ============================================

# Extract only date from datetime

activity["date"] = activity["activity_date"].dt.date

sleep["date"] = sleep["sleep_day"].dt.date

weight["date"] = weight["date"].dt.date

heart["date"] = heart["time"].dt.date

print("✅ Date columns prepared successfully!")

print(activity[["id","date"]].head())

print(sleep[["id","date"]].head())

print(weight[["id","date"]].head())

print(heart[["id","date"]].head())
```

```python
# ============================================
# Feature Engineering - Step 8
# Daily Heart Rate Summary
# ============================================

heart_daily = (
    heart.groupby(["id", "date"])
         .agg(
             avg_heart_rate=("value", "mean"),
             min_heart_rate=("value", "min"),
             max_heart_rate=("value", "max")
         )
         .reset_index()
)

heart_daily["avg_heart_rate"] = heart_daily["avg_heart_rate"].round(2)

print("✅ Daily Heart Rate Summary Created Successfully!\n")

print(heart_daily.head())

print("\nShape:", heart_daily.shape)
```

```python
# ============================================
# Feature Engineering - Step 9
# Merge Fitbit Datasets
# ============================================

fitbit_data = activity.merge(
    sleep,
    on=["id", "date"],
    how="left"
)

fitbit_data = fitbit_data.merge(
    weight,
    on=["id", "date"],
    how="left"
)

fitbit_data = fitbit_data.merge(
    heart_daily,
    on=["id", "date"],
    how="left"
)

print("✅ Fitbit Master Dataset Created Successfully!\n")

print("Shape:", fitbit_data.shape)

print("\nColumns:\n")

print(fitbit_data.columns.tolist())

print("\nFirst 5 Rows:\n")

print(fitbit_data.head())
```

```python
# ============================================
# MACHINE LEARNING
# Step 1 : Select Features & Target
# ============================================

# Features (Input)
X = body.drop("class", axis=1)

# Target (Output)
y = body["class"]

print("✅ Features and Target Selected Successfully!\n")

print("Features Shape :", X.shape)
print("Target Shape   :", y.shape)

print("\nFeature Columns:\n")
print(X.columns.tolist())

print("\nTarget Classes:\n")
print(y.value_counts())
```

```python
# ============================================
# MACHINE LEARNING
# Step 2 : Encode Categorical Variables
# ============================================

from sklearn.preprocessing import LabelEncoder

# Create Label Encoders
gender_encoder = LabelEncoder()
bp_encoder = LabelEncoder()
class_encoder = LabelEncoder()

# Encode Features
X["gender"] = gender_encoder.fit_transform(X["gender"])
X["bp_category"] = bp_encoder.fit_transform(X["bp_category"])

# Encode Target
y = class_encoder.fit_transform(y)

print("✅ Categorical Variables Encoded Successfully!\n")

print("Encoded Gender:")
print(dict(zip(gender_encoder.classes_,
               gender_encoder.transform(gender_encoder.classes_))))

print("\nEncoded BP Category:")
print(dict(zip(bp_encoder.classes_,
               bp_encoder.transform(bp_encoder.classes_))))

print("\nEncoded Target Classes:")
print(dict(zip(class_encoder.classes_,
               class_encoder.transform(class_encoder.classes_))))
```

```python
print("\nFirst 5 Rows of Features:\n")
print(X.head())

print("\nFirst 10 Target Values:\n")
print(y[:10])
```

```python
# ============================================
# MACHINE LEARNING
# Step 3 : Train-Test Split
# ============================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("✅ Train-Test Split Completed Successfully!\n")

print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print()

print("Training Target :", y_train.shape)
print("Testing Target  :", y_test.shape)
```

```python
# ============================================
# MACHINE LEARNING
# Step 4 : Feature Scaling
# ============================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

# Numerical columns
numerical_columns = [
    "age",
    "height_cm",
    "weight_kg",
    "body_fat_percent",
    "diastolic",
    "systolic",
    "grip_force",
    "sit_and_bend_forward_cm",
    "sit_ups_counts",
    "broad_jump_cm"
]

# Apply scaling only to numerical columns
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns)
    ],
    remainder="passthrough"  # Keep encoded categorical columns unchanged
)

X_train_scaled = preprocessor.fit_transform(X_train)
X_test_scaled = preprocessor.transform(X_test)

print("✅ Feature Scaling Completed Successfully!\n")

print("Scaled Training Shape :", X_train_scaled.shape)
print("Scaled Testing Shape  :", X_test_scaled.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create Model
dt_model = DecisionTreeClassifier(random_state=42)

# Train Model
dt_model.fit(X_train, y_train)

# Prediction
y_pred = dt_model.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("="*60)
print("Decision Tree Results")
print("="*60)

print(f"Accuracy : {accuracy:.4f}")

print("\nClassification Report\n")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test, y_pred))
```

```python
# ============================================
# MACHINE LEARNING
# Step 6 : Random Forest Classifier
# ============================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create Model
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

# Train
rf_model.fit(X_train, y_train)

# Prediction
y_pred_rf = rf_model.predict(X_test)

# Accuracy
rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("="*60)
print("Random Forest Results")
print("="*60)

print(f"Accuracy : {rf_accuracy:.4f}")

print("\nClassification Report\n")
print(classification_report(y_test, y_pred_rf))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test, y_pred_rf))
```

```python
# ==========================================================
# Soldier Readiness - Individual Scores
# ==========================================================

import numpy as np

# ---------------- Activity Score ----------------

fitbit_data["activity_score"] = np.select(
    [
        fitbit_data["total_steps"] >= 10000,
        fitbit_data["total_steps"] >= 7500,
        fitbit_data["total_steps"] >= 5000,
        fitbit_data["total_steps"] >= 2500
    ],
    [100, 80, 60, 40],
    default=20
)

# ---------------- Sleep Score ----------------

fitbit_data["sleep_score"] = np.where(
    fitbit_data["sleep_efficiency"].isna(),
    np.nan,
    np.select(
        [
            fitbit_data["sleep_efficiency"] >= 95,
            fitbit_data["sleep_efficiency"] >= 90,
            fitbit_data["sleep_efficiency"] >= 85,
            fitbit_data["sleep_efficiency"] >= 80
        ],
        [100, 80, 60, 40],
        default=20
    )
)

# ---------------- Heart Score ----------------

fitbit_data["heart_score"] = np.where(
    fitbit_data["avg_heart_rate"].isna(),
    np.nan,
    np.where(
        fitbit_data["avg_heart_rate"].between(60,90),
        100,
        np.where(
            fitbit_data["avg_heart_rate"].between(50,100),
            80,
            60
        )
    )
)

# ---------------- BMI Score ----------------

fitbit_data["bmi_score"] = np.where(
    fitbit_data["bmi"].isna(),
    np.nan,
    np.where(
        fitbit_data["bmi"].between(18.5,24.9),
        100,
        np.where(
            fitbit_data["bmi"].between(25,29.9),
            70,
            50
        )
    )
)

print("✅ Individual Scores Created Successfully!\n")

print(
    fitbit_data[
        [
            "activity_score",
            "sleep_score",
            "heart_score",
            "bmi_score"
        ]
    ].head()
)
```

```python
# ==========================================================
# Final Soldier Readiness Score
# ==========================================================

# Calculate average of available scores (NaN ignored automatically)

fitbit_data["soldier_readiness_score"] = fitbit_data[
    [
        "activity_score",
        "sleep_score",
        "heart_score",
        "bmi_score"
    ]
].mean(axis=1)

# Create Readiness Status

fitbit_data["readiness_status"] = np.select(
    [
        fitbit_data["soldier_readiness_score"] >= 90,
        fitbit_data["soldier_readiness_score"] >= 75,
        fitbit_data["soldier_readiness_score"] >= 60,
    ],
    [
        "Excellent",
        "Fit for Duty",
        "Needs Improvement"
    ],
    default="Critical"
)

print("✅ Soldier Readiness Score Created Successfully!\n")

print(
    fitbit_data[
        [
            "activity_score",
            "sleep_score",
            "heart_score",
            "bmi_score",
            "soldier_readiness_score",
            "readiness_status"
        ]
    ].head(10)
)

print("\nReadiness Status Distribution:\n")
print(fitbit_data["readiness_status"].value_counts())
```
