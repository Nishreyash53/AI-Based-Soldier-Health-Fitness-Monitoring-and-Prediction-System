```python
# ============================================
# Feature Engineering - Step 1
# BMI Category
# ============================================

def bmi_category(bmi):

    if bmi < 18.5:
        return "Underweight"

    elif bmi < 25:
        return "Normal"

    elif bmi < 30:
        return "Overweight"

    else:
        return "Obese"


weight["bmi_category"] = weight["bmi"].apply(bmi_category)

print("✅ BMI Category Created Successfully!\n")

print(weight[["bmi", "bmi_category"]].head())
```

```python
# ============================================
# Feature Engineering - Step 2
# Blood Pressure Category
# ============================================

def bp_category(systolic, diastolic):

    if systolic < 120 and diastolic < 80:
        return "Normal"

    elif 120 <= systolic < 130 and diastolic < 80:
        return "Elevated"

    elif (130 <= systolic < 140) or (80 <= diastolic < 90):
        return "Hypertension Stage 1"

    elif systolic >= 140 or diastolic >= 90:
        return "Hypertension Stage 2"

    else:
        return "Unknown"


body["bp_category"] = body.apply(
    lambda row: bp_category(row["systolic"], row["diastolic"]),
    axis=1
)

print("✅ Blood Pressure Category Created Successfully!\n")

print(body[["systolic", "diastolic", "bp_category"]].head())
```

```python
print(body["bp_category"].value_counts())
```

```python
# ============================================
# Feature Engineering - Step 3
# Sleep Efficiency
# ============================================

sleep["sleep_efficiency"] = (
    sleep["total_minutes_asleep"] / sleep["total_time_in_bed"]
) * 100

sleep["sleep_efficiency"] = sleep["sleep_efficiency"].round(2)

print("✅ Sleep Efficiency Created Successfully!\n")

print(
    sleep[
        [
            "total_minutes_asleep",
            "total_time_in_bed",
            "sleep_efficiency"
        ]
    ].head()
)
```

```python
# ============================================
# Sleep Quality Category
# ============================================

def sleep_quality(efficiency):

    if efficiency >= 90:
        return "Excellent"

    elif efficiency >= 85:
        return "Good"

    elif efficiency >= 75:
        return "Average"

    else:
        return "Poor"


sleep["sleep_quality"] = sleep["sleep_efficiency"].apply(sleep_quality)

print("\n✅ Sleep Quality Category Created!\n")

print(
    sleep[
        [
            "sleep_efficiency",
            "sleep_quality"
        ]
    ].head()
)
```

```python
# ============================================
# Feature Engineering - Step 4
# Activity Level
# ============================================

def activity_level(steps):

    if steps < 5000:
        return "Sedentary"

    elif steps < 7500:
        return "Low Active"

    elif steps < 10000:
        return "Moderately Active"

    else:
        return "Highly Active"


activity["activity_level"] = activity["total_steps"].apply(activity_level)

print("✅ Activity Level Created Successfully!\n")

print(activity[["total_steps", "activity_level"]].head())

print("\nActivity Level Distribution:\n")

print(activity["activity_level"].value_counts())
```

```python
# ============================================
# Feature Engineering - Step 5
# Heart Rate Summary
# ============================================

heart_summary = (
    heart.groupby("id")
         .agg(
             avg_heart_rate=("value", "mean"),
             min_heart_rate=("value", "min"),
             max_heart_rate=("value", "max")
         )
         .reset_index()
)

# Round average heart rate
heart_summary["avg_heart_rate"] = heart_summary["avg_heart_rate"].round(2)

print("✅ Heart Rate Summary Created Successfully!\n")

print(heart_summary.head())

print("\nShape of Heart Summary Dataset:")
print(heart_summary.shape)
```

```python
# ============================================
# Feature Engineering - Step 6
# Heart Rate Category
# ============================================

def heart_rate_category(avg_hr):

    if avg_hr < 60:
        return "Low"

    elif avg_hr <= 100:
        return "Normal"

    else:
        return "High"


heart_summary["heart_rate_category"] = heart_summary["avg_heart_rate"].apply(
    heart_rate_category
)

print("✅ Heart Rate Category Created Successfully!\n")

print(heart_summary)

print("\nCategory Distribution:\n")

print(heart_summary["heart_rate_category"].value_counts())
```

```python
# ============================================
# Feature Engineering - Step 7
# Prepare Fitbit Datasets for Merge
# ============================================

# Extract only date from datetime

activity["date"] = activity["activity_date"].dt.date

sleep["date"] = sleep["sleep_day"].dt.date

weight["date"] = weight["date"].dt.date

heart["date"] = heart["time"].dt.date

print("✅ Date columns prepared successfully!")

print(activity[["id","date"]].head())

print(sleep[["id","date"]].head())

print(weight[["id","date"]].head())

print(heart[["id","date"]].head())
```

```python
# ============================================
# Feature Engineering - Step 8
# Daily Heart Rate Summary
# ============================================

heart_daily = (
    heart.groupby(["id", "date"])
         .agg(
             avg_heart_rate=("value", "mean"),
             min_heart_rate=("value", "min"),
             max_heart_rate=("value", "max")
         )
         .reset_index()
)

heart_daily["avg_heart_rate"] = heart_daily["avg_heart_rate"].round(2)

print("✅ Daily Heart Rate Summary Created Successfully!\n")

print(heart_daily.head())

print("\nShape:", heart_daily.shape)
```

```python
# ============================================
# Feature Engineering - Step 9
# Merge Fitbit Datasets
# ============================================

fitbit_data = activity.merge(
    sleep,
    on=["id", "date"],
    how="left"
)

fitbit_data = fitbit_data.merge(
    weight,
    on=["id", "date"],
    how="left"
)

fitbit_data = fitbit_data.merge(
    heart_daily,
    on=["id", "date"],
    how="left"
)

print("✅ Fitbit Master Dataset Created Successfully!\n")

print("Shape:", fitbit_data.shape)

print("\nColumns:\n")

print(fitbit_data.columns.tolist())

print("\nFirst 5 Rows:\n")

print(fitbit_data.head())
```

```python
# ============================================
# MACHINE LEARNING
# Step 1 : Select Features & Target
# ============================================

# Features (Input)
X = body.drop("class", axis=1)

# Target (Output)
y = body["class"]

print("✅ Features and Target Selected Successfully!\n")

print("Features Shape :", X.shape)
print("Target Shape   :", y.shape)

print("\nFeature Columns:\n")
print(X.columns.tolist())

print("\nTarget Classes:\n")
print(y.value_counts())
```

```python
# ============================================
# MACHINE LEARNING
# Step 2 : Encode Categorical Variables
# ============================================

from sklearn.preprocessing import LabelEncoder

# Create Label Encoders
gender_encoder = LabelEncoder()
bp_encoder = LabelEncoder()
class_encoder = LabelEncoder()

# Encode Features
X["gender"] = gender_encoder.fit_transform(X["gender"])
X["bp_category"] = bp_encoder.fit_transform(X["bp_category"])

# Encode Target
y = class_encoder.fit_transform(y)

print("✅ Categorical Variables Encoded Successfully!\n")

print("Encoded Gender:")
print(dict(zip(gender_encoder.classes_,
               gender_encoder.transform(gender_encoder.classes_))))

print("\nEncoded BP Category:")
print(dict(zip(bp_encoder.classes_,
               bp_encoder.transform(bp_encoder.classes_))))

print("\nEncoded Target Classes:")
print(dict(zip(class_encoder.classes_,
               class_encoder.transform(class_encoder.classes_))))
```

```python
print("\nFirst 5 Rows of Features:\n")
print(X.head())

print("\nFirst 10 Target Values:\n")
print(y[:10])
```

```python
# ============================================
# MACHINE LEARNING
# Step 3 : Train-Test Split
# ============================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("✅ Train-Test Split Completed Successfully!\n")

print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print()

print("Training Target :", y_train.shape)
print("Testing Target  :", y_test.shape)
```

```python
# ============================================
# MACHINE LEARNING
# Step 4 : Feature Scaling
# ============================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

# Numerical columns
numerical_columns = [
    "age",
    "height_cm",
    "weight_kg",
    "body_fat_percent",
    "diastolic",
    "systolic",
    "grip_force",
    "sit_and_bend_forward_cm",
    "sit_ups_counts",
    "broad_jump_cm"
]

# Apply scaling only to numerical columns
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns)
    ],
    remainder="passthrough"  # Keep encoded categorical columns unchanged
)

X_train_scaled = preprocessor.fit_transform(X_train)
X_test_scaled = preprocessor.transform(X_test)

print("✅ Feature Scaling Completed Successfully!\n")

print("Scaled Training Shape :", X_train_scaled.shape)
print("Scaled Testing Shape  :", X_test_scaled.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create Model
dt_model = DecisionTreeClassifier(random_state=42)

# Train Model
dt_model.fit(X_train, y_train)

# Prediction
y_pred = dt_model.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("="*60)
print("Decision Tree Results")
print("="*60)

print(f"Accuracy : {accuracy:.4f}")

print("\nClassification Report\n")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test, y_pred))
```

```python
# ============================================
# MACHINE LEARNING
# Step 6 : Random Forest Classifier
# ============================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create Model
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

# Train
rf_model.fit(X_train, y_train)

# Prediction
y_pred_rf = rf_model.predict(X_test)

# Accuracy
rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("="*60)
print("Random Forest Results")
print("="*60)

print(f"Accuracy : {rf_accuracy:.4f}")

print("\nClassification Report\n")
print(classification_report(y_test, y_pred_rf))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test, y_pred_rf))
```

```python
# ==========================================================
# Soldier Readiness - Individual Scores
# ==========================================================

import numpy as np

# ---------------- Activity Score ----------------

fitbit_data["activity_score"] = np.select(
    [
        fitbit_data["total_steps"] >= 10000,
        fitbit_data["total_steps"] >= 7500,
        fitbit_data["total_steps"] >= 5000,
        fitbit_data["total_steps"] >= 2500
    ],
    [100, 80, 60, 40],
    default=20
)

# ---------------- Sleep Score ----------------

fitbit_data["sleep_score"] = np.where(
    fitbit_data["sleep_efficiency"].isna(),
    np.nan,
    np.select(
        [
            fitbit_data["sleep_efficiency"] >= 95,
            fitbit_data["sleep_efficiency"] >= 90,
            fitbit_data["sleep_efficiency"] >= 85,
            fitbit_data["sleep_efficiency"] >= 80
        ],
        [100, 80, 60, 40],
        default=20
    )
)

# ---------------- Heart Score ----------------

fitbit_data["heart_score"] = np.where(
    fitbit_data["avg_heart_rate"].isna(),
    np.nan,
    np.where(
        fitbit_data["avg_heart_rate"].between(60,90),
        100,
        np.where(
            fitbit_data["avg_heart_rate"].between(50,100),
            80,
            60
        )
    )
)

# ---------------- BMI Score ----------------

fitbit_data["bmi_score"] = np.where(
    fitbit_data["bmi"].isna(),
    np.nan,
    np.where(
        fitbit_data["bmi"].between(18.5,24.9),
        100,
        np.where(
            fitbit_data["bmi"].between(25,29.9),
            70,
            50
        )
    )
)

print("✅ Individual Scores Created Successfully!\n")

print(
    fitbit_data[
        [
            "activity_score",
            "sleep_score",
            "heart_score",
            "bmi_score"
        ]
    ].head()
)
```

```python
import pandas as pd

path = r"N:\NISHREYASH\software_eneg,ml,dv\main_datset"

body = pd.read_csv(path + r"\bodyPerformance.csv")

print(body.shape)
```

```python
# ======================================================
# MASTER DATA PREPARATION
# ======================================================

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

# Features & Target
X = body.drop("class", axis=1)
y = body["class"]

# Encode Gender
gender_encoder = LabelEncoder()
X["gender"] = gender_encoder.fit_transform(X["gender"])

# Encode Target
class_encoder = LabelEncoder()
y = class_encoder.fit_transform(y)

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Numerical Columns
numerical_columns = [
    "age",
    "height_cm",
    "weight_kg",
    "body fat_%",
    "diastolic",
    "systolic",
    "gripForce",
    "sit and bend forward_cm",
    "sit-ups counts",
    "broad jump_cm"
]

# Preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns)
    ],
    remainder="passthrough"
)

# Transform
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("✅ Data Preparation Completed")
print("Train Shape :", X_train_processed.shape)
print("Test Shape  :", X_test_processed.shape)
```

```python
# ======================================================
# TRAIN OPTUNA TUNED XGBOOST
# ======================================================

from xgboost import XGBClassifier

best_xgb = XGBClassifier(
    n_estimators=363,
    max_depth=7,
    learning_rate=0.03724913888151351,
    subsample=0.9997180951904389,
    colsample_bytree=0.8611199450966237,
    gamma=0.06308301783025762,
    min_child_weight=10,
    reg_alpha=0.7822777877158005,
    reg_lambda=6.958745822661207,
    objective="multi:softprob",
    num_class=4,
    eval_metric="mlogloss",
    random_state=42
)

best_xgb.fit(X_train_processed, y_train)

print("✅ Tuned XGBoost Model Trained Successfully")
```

```python
# ======================================================
# EVALUATE TUNED XGBOOST
# ======================================================

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = best_xgb.predict(X_test_processed)

accuracy = accuracy_score(y_test, y_pred)

print("="*60)
print("🚀 TUNED XGBOOST RESULTS")
print("="*60)

print(f"Accuracy : {accuracy*100:.2f}%")

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred))
```

```python
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, classification_report

# Original data use karenge (processed/scaled nahi)
X = body.drop("class", axis=1)
y = body["class"]

# Encode target only
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)

# Train/Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# CatBoost ko batana hai kaunsa column categorical hai
cat_features = [X.columns.get_loc("gender")]

cat_model = CatBoostClassifier(
    iterations=500,
    learning_rate=0.05,
    depth=8,
    loss_function="MultiClass",
    eval_metric="Accuracy",
    random_seed=42,
    verbose=100
)

cat_model.fit(
    X_train,
    y_train,
    cat_features=cat_features
)

# Prediction
y_pred = cat_model.predict(X_test)

print("="*60)
print("CATBOOST RESULTS")
print("="*60)

print("Accuracy :", accuracy_score(y_test, y_pred))

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

```python
# ======================================================
# EVALUATE STACKING MODEL
# ======================================================

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred_stack = stack_model.predict(X_test_processed)

accuracy = accuracy_score(y_test, y_pred_stack)

print("="*60)
print("🚀 STACKING ENSEMBLE RESULTS")
print("="*60)

print(f"Accuracy : {accuracy*100:.2f}%")

print("\nClassification Report\n")
print(classification_report(y_test, y_pred_stack))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test, y_pred_stack))
```

```python
# ============================================
# FEATURE ENGINEERING
# ============================================

body["BMI"] = body["weight_kg"] / ((body["height_cm"] / 100) ** 2)

body["PulsePressure"] = body["systolic"] - body["diastolic"]

body["StrengthIndex"] = body["gripForce"] / body["weight_kg"]

body["JumpPower"] = body["broad jump_cm"] / body["weight_kg"]

body["CoreStrength"] = body["sit-ups counts"] / body["weight_kg"]

body["FlexibilityIndex"] = body["sit and bend forward_cm"] / body["height_cm"]

print(body.head())
```

```python
# ===========================================
# STEP 1 : Create New Features & Target
# ===========================================

X = body.drop("class", axis=1)
y = body["class"]

print("New Feature Shape :", X.shape)
print("Target Shape :", y.shape)
```

```python
# ===========================================
# STEP 2 : Encode + Train/Test Split
# ===========================================

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

# Encode Gender
gender_encoder = LabelEncoder()
X["gender"] = gender_encoder.fit_transform(X["gender"])

# Encode Target
class_encoder = LabelEncoder()
y = class_encoder.fit_transform(y)

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train Shape :", X_train.shape)
print("Test Shape  :", X_test.shape)
```

```python
# ===========================================
# STEP 3 : Updated Preprocessing
# ===========================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

numerical_columns = [
    "age",
    "height_cm",
    "weight_kg",
    "body fat_%",
    "diastolic",
    "systolic",
    "gripForce",
    "sit and bend forward_cm",
    "sit-ups counts",
    "broad jump_cm",

    # Engineered Features
    "BMI",
    "PulsePressure",
    "StrengthIndex",
    "JumpPower",
    "CoreStrength",
    "FlexibilityIndex"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns)
    ],
    remainder="passthrough"   # gender
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Train Shape :", X_train_processed.shape)
print("Test Shape  :", X_test_processed.shape)
```

```python
# ===========================================
# STEP 4 : Retrain Tuned XGBoost
# ===========================================

from xgboost import XGBClassifier

best_xgb = XGBClassifier(
    n_estimators=363,
    max_depth=7,
    learning_rate=0.03724913888151351,
    subsample=0.9997180951904389,
    colsample_bytree=0.8611199450966237,
    gamma=0.06308301783025762,
    min_child_weight=10,
    reg_alpha=0.7822777877158005,
    reg_lambda=6.958745822661207,
    objective="multi:softprob",
    num_class=4,
    eval_metric="mlogloss",
    random_state=42
)

best_xgb.fit(X_train_processed, y_train)

print("✅ Retraining Completed")
```

```python
# ===========================================
# STEP 5 : Evaluate New XGBoost
# ===========================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Prediction
y_pred = best_xgb.predict(X_test_processed)

# Metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average="weighted")
recall = recall_score(y_test, y_pred, average="weighted")
f1 = f1_score(y_test, y_pred, average="weighted")

print("="*60)
print("🚀 XGBOOST + FEATURE ENGINEERING RESULTS")
print("="*60)

print(f"Accuracy  : {accuracy*100:.2f}%")
print(f"Precision : {precision*100:.2f}%")
print(f"Recall    : {recall*100:.2f}%")
print(f"F1 Score  : {f1*100:.2f}%")

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred))
```

```python
# ==========================================
# ADVANCED OPTUNA OBJECTIVE (v2)
# ==========================================

from sklearn.model_selection import cross_val_score
from xgboost import XGBClassifier

def objective(trial):

    params = {

        "n_estimators": trial.suggest_int("n_estimators", 200, 700),

        "max_depth": trial.suggest_int("max_depth", 3, 10),

        "learning_rate": trial.suggest_float(
            "learning_rate",
            0.01,
            0.20,
            log=True
        ),

        "min_child_weight": trial.suggest_int(
            "min_child_weight",
            1,
            15
        ),

        "subsample": trial.suggest_float(
            "subsample",
            0.6,
            1.0
        ),

        "colsample_bytree": trial.suggest_float(
            "colsample_bytree",
            0.6,
            1.0
        ),

        "colsample_bylevel": trial.suggest_float(
            "colsample_bylevel",
            0.6,
            1.0
        ),

        "gamma": trial.suggest_float(
            "gamma",
            0,
            5
        ),

        "reg_alpha": trial.suggest_float(
            "reg_alpha",
            0,
            10
        ),

        "reg_lambda": trial.suggest_float(
            "reg_lambda",
            1,
            15
        ),

        "max_delta_step": trial.suggest_int(
            "max_delta_step",
            0,
            10
        ),

        "objective": "multi:softprob",

        "num_class": 4,

        "eval_metric": "mlogloss",

        "random_state": 42,

        "tree_method": "hist",

        "n_jobs": -1
    }

    model = XGBClassifier(**params)

    score = cross_val_score(
        model,
        X_train_processed,
        y_train,
        cv=5,
        scoring="accuracy",
        n_jobs=-1
    )

    return score.mean()

print("✅ Advanced Objective Ready")
```

```python
# ==========================================
# CREATE OPTUNA STUDY
# ==========================================

import optuna

study = optuna.create_study(
    direction="maximize",
    sampler=optuna.samplers.TPESampler(seed=42)
)

study.optimize(
    objective,
    n_trials=50,
    show_progress_bar=True
)

print("="*60)
print("BEST CV ACCURACY :", study.best_value)
print("="*60)

print("\nBEST PARAMETERS\n")
print(study.best_params)
```
